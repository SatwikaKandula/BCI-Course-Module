# Step 1: Load and Visualize EEG Data
This notebook loads raw EEG data from the PhysioNet Motor Imagery dataset
and visualizes the brain signals.

```python
import mne
```

```python
# Load one subject file
raw = mne.io.read_raw_edf('../DATA/S001R03.edf', preload=True)
# Print basic info
print(raw.info)
```

```python
# How many channels and what are they?
print("Number of channels:", len(raw.ch_names))
print("Channel names:", raw.ch_names)
print("Sampling frequency:", raw.info['sfreq'], "Hz")
print("Duration:", raw.times[-1], "seconds")
```

```python
# Pick just a few channels to plot (easier to see)
channels_to_plot = ['C3..', 'C4..', 'Cz..']
# Plot 10 seconds of raw EEG
raw.plot(
    duration=10,
    n_channels=3,
    picks=channels_to_plot,
    title='Raw EEG Signal — Subject 1',
    show=True,
)
channels_to_plot = ['C3..', 'C4..', 'Cz..']
```

Note: The EEG plot opens as an interactive MNE window when the notebook is run locally. GitHub may not display this interactive plot in the notebook preview.
## What are we seeing?
- Black lines = raw brain signals from 3 electrodes
- Blue = rest (T0), Green = motor imagery task (T1/T2)
- Signal looks noisy — this is why we need filtering next

```python
# Make a copy so we keep the original raw signal
raw_filtered = raw.copy()
# Apply band-pass filter — keep only 8 to 30 Hz
raw_filtered.filter(8., 30., fir_design='firwin')
print("Filtering done!")
```

```python
# Plot original (noisy)
raw.plot(
    duration=10,
    n_channels=3,
    picks=channels_to_plot,
    title='BEFORE Filtering — Raw EEG',
    show=True,
)

# Plot filtered
raw_filtered.plot(
    duration=10,
    n_channels=3,
    picks=channels_to_plot,
    title='AFTER Filtering — 8-30 Hz',
    show=True,
)
```
